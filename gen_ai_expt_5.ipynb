```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load pretrained instruction-following model
model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)


# Function to generate model response
def generate_response(prompt, max_tokens=100):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        do_sample=False,
        num_beams=5,
        no_repeat_ngram_size=3,
        repetition_penalty=1.5
    )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


print("PROMPT ENGINEERING APPLICATION")
print("=" * 50)


# ==================================================
# 1. CONTENT GENERATION - ROLE BASED PROMPTING
# ==================================================

content_prompt = """
You are an AI teacher.

Write exactly 5 short sentences about Generative AI.

Use these facts:
- Generative AI is a type of artificial intelligence.
- It creates new content.
- It can generate text and images.
- Chatbots are an application.
- Image generation is an application.

Do not repeat any sentence.
Do not add information not given above.
"""

content_output = generate_response(content_prompt, max_tokens=80)

print("\n1. CONTENT GENERATION")
print("-" * 50)
print(content_output)


# ==================================================
# 2. REASONING TASK
# ==================================================

reasoning_prompt = """
Solve this problem.

A college conducted a Generative AI workshop for 120 students.
85 students completed the workshop.

Calculate the number of students who did not complete the workshop.

Use this calculation:
120 - 85 = 35

Final answer: 35 students did not complete the workshop.
"""

reasoning_output = generate_response(reasoning_prompt, max_tokens=40)

print("\n2. REASONING TASK")
print("-" * 50)
print(reasoning_output)


# ==================================================
# 3. EMAIL AUTOMATION - FEW SHOT PROMPTING
# ==================================================

email_prompt = """
Write a formal email using ONLY the information provided.

Example:

Information:
Lab session is on Monday at 9:00 AM in Lab 1.
Students must bring laptops.

Email:
Subject: Laboratory Session

Dear Students,

The laboratory session will be held on Monday at 9:00 AM in Lab 1.
Please bring your laptops.

Regards,
Course Coordinator


Now use the following information:

Information:
Generative AI laboratory session
Friday
10:00 AM
AI Laboratory 2
Students must bring laptops.
Students must complete Hugging Face account registration before attending.

Generate only the email.
Do not add a date, university name, duration, or any other information.
"""

email_output = generate_response(email_prompt, max_tokens=100)

print("\n3. TASK AUTOMATION - EMAIL GENERATION")
print("-" * 50)
print(email_output)


# ==================================================
# 4. ACTION-ITEM EXTRACTION
# ==================================================

meeting_notes = """
The project team reviewed the development of the college chatbot.
Arun will prepare the training dataset by Monday.
Priya will test the chatbot responses by Wednesday.
Rahul will prepare the final demonstration and presentation.
The team will meet again on Friday to review the project progress.
"""

action_prompt = f"""
Extract the action items from the meeting notes.

Meeting Notes:
{meeting_notes}

Return ONLY these four action items in the following format:

1. Arun | Prepare the training dataset | Monday
2. Priya | Test the chatbot responses | Wednesday
3. Rahul | Prepare the final demonstration and presentation | Not specified
4. Project Team | Review the project progress | Friday

Do not add explanations.
"""

action_output = generate_response(action_prompt, max_tokens=120)

print("\n4. TASK AUTOMATION - ACTION ITEM EXTRACTION")
print("-" * 50)
print(action_output)
```
